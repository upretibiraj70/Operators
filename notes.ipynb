```python
#airthmetic operator
#+-*/-add,subtract,multiple,divide
#%-modulo
#**-power
#//floor division
a=5
b=2
print(a+b) #7
print(a-b) #3
print(a*b) #10
print(a/b) #2.5
print(a%b) #1
print(a//b) #2
print(a**b) #25
```

```python
#comparision operators
#==equal
#!=not equal
#>greater than
#<less than
#>= greater or equal
#<= less or equal
print(10>5) #true
print(10==5) #False
print(10>5)#true
print(10<5)#False
print (10!=5)#true
print(10>=5)#true
print(10<=5)#False
```

```python
#Assignment operator
#= equal
"""= equal
+= Assign
-= Subtract assign
*= multiple assign
/= Divide assign
//= Floor division assign
%= Modulus assign
**= Power assign"""
score = 50
score += 10
print(score)
score *= 2 
print(score)
score -= 20
print(score)
score /= 10
print(score)
score %= 10
print(score)
score //= 10
print(score)
score **= 10
print(score)
```

```python
#logical operator
""""and
or
not"""
marks = 72
print(marks >= 40 and marks <= 100) 
print(marks < 40 or marks > 100)
```

```python
#Bitwise operator
""""& and
| or
^ XOR
~ NOT
<< Left shift
>> right shift
"""
print(5 & 3)
print(5 | 3)
print(5 ^ 3)
print(5 << 3)
print(5 >> 3)
```

```python
#identity operator
""""is
is not"""
a=[1,2,3]
b=[1,2,3]
c=a
print(a==b)
print(a is not b)
print(a is c)
print(id(a))
print(id(b))
print(id(c))
```
